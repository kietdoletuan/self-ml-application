```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Read dataset

```python
dataset_path = "data/opsd_germany_daily.csv"

opsd_daily = pd.read_csv(dataset_path, index_col=0, parse_dates=True)
```

```python
opsd_daily["Year"] = opsd_daily.index.year
opsd_daily["Month"] = opsd_daily.index.month
opsd_daily["Weekday Name"] = opsd_daily.index.day_name()


opsd_daily.sample(5, random_state=0)
```

## Visualization

```python
sns.set(rc={"figure.figsize":(11, 4)})
col_to_plot = 'Consumption'
opsd_daily[col_to_plot].plot(linewidth=0.5)
plt.ylabel(col_to_plot)
plt.show()
```

```python
cols_plot = ["Consumption", "Solar", "Wind"]
axes = opsd_daily[cols_plot].plot(
    marker='.',
    linestyle ='None',
    figsize=(10, 12),
    subplots=True
)

for ax in axes:
    ax.set_ylabel('Daily Totals (GWh)')

plt.show
```

## Seasonality

```python
fig, axes = plt.subplots(3, 1, figsize=(11, 10), sharex=True)

for name, ax in zip(["Consumption", "Solar", "Wind"], axes):
    sns.boxplot(data=opsd_daily, x="Month", y=name, ax=ax)
    ax.set_ylabel('GWh')
    ax.set_title(name)

    

```

## Frequencies

```python
times_sample = pd.to_datetime(["2013-02-03", "2013-02-06", "2013-02-08"])
consum_sample = opsd_daily.loc[times_sample, ['Consumption']].copy()
consum_sample
```

```python
consum_freq = consum_sample.asfreq('D')

consum_freq["Consumption - Forward Fill"] = consum_sample.asfreq("D", method="ffill")
consum_freq
```

```python
data_columns = ["Consumption", "Wind", "Solar", "Wind+Solar"]
opsd_weekly_mean = opsd_daily[data_columns].resample('W').mean()
opsd_weekly_mean.head()
```

```python
print(opsd_daily.shape[0])
print(opsd_weekly_mean.shape[0])
```

```python
start, end = "2017-01", "2017-06"

fig, ax = plt.subplots()
ax.plot(opsd_daily.loc[start:end, "Solar"],
        marker='.',
        linestyle='-',
        linewidth=0.5,
        label='Daily')
ax.plot(opsd_weekly_mean.loc[start:end, "Solar"],
        marker='o',
        markersize=8,
        linestyle='-',
        label='Weekly Mean Resample')
ax.set_ylabel("Solar Production (GWh)")
ax.legend()
plt.show()
```

```python
opsd_annual = opsd_daily[data_columns].resample('YE').sum(min_count=360)
opsd_annual = opsd_annual.set_index(opsd_annual.index.year)
opsd_annual.index.name = 'Year'

opsd_annual['Wind+Solar/Consumption'] = opsd_annual['Wind+Solar'] / opsd_annual['Consumption']
opsd_annual.tail()
```

```python
ax = opsd_annual.loc[2012:, 'Wind+Solar/Consumption'].plot.bar(color='C0')
ax.set_ylabel('Fraction')
ax.set_ylim(0, 0.3)
ax.set_title("Wind + Solar Share of Annual Electricity Consumption")
plt.xticks(rotation=0)
plt.show()
```

## Rolling Windows

```python
opsd_7d = opsd_daily[data_columns].rolling(7, center=True).mean()
opsd_7d.head(10)
```

## Trends

```python
import matplotlib.dates as mdates
```

```python
opsd_365d = opsd_daily[data_columns].rolling(
    window=365,
    center=True,
    min_periods=360
).mean()

fig, ax = plt.subplots()
ax.plot(opsd_daily["Consumption"], marker=".", markersize=2, color="0.6", linestyle="None", label="Daily")

ax.plot(opsd_7d["Consumption"], linewidth=2, label="7-d Rolling Mean")

ax.plot(opsd_365d["Consumption"], color="0.2", linewidth=3, label="Trend (365-d Rolling Mean)")

ax.xaxis.set_major_locator(mdates.YearLocator())
ax.legend()
ax.set_xlabel("Year")
ax.set_ylabel("Consumption (GWh)")
ax.set_title("Trends in Electricity Consumption")
plt.show()
```

```python
fig, ax = plt.subplots()
for nm in ["Wind", "Solar", "Wind+Solar"]:
    ax.plot(opsd_365d[nm], label=nm)
    ax.xaxis.set_major_locator(mdates.YearLocator())
    ax.set_ylim(0, 400)
    ax.legend()
    ax.set_ylabel("Production (GWh)")
    ax.set_title("Trends in Electricity Production (365-d Rolling Means)")
plt.show()
```
