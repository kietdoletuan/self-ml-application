```python
import pandas as pd
```

## Read Data

```python
dataset_path = "data/IMDB-Movie-Data.csv"
data = pd.read_csv(dataset_path)
data_indexed = pd.read_csv(dataset_path, index_col='Title')
```

## Data Basic Information

```python
data.head()
```

```python
data.info()
```

## Data Selection

```python
get_columns = data[['Title', 'Rating', 'Revenue (Millions)']][10:15]
get_columns
```

## Data Selection - Conditional Filter

```python
get_data = data[(data['Year'].isin(range(2010,2016))) & (data['Rating'] < 6) & (data['Revenue (Millions)'] >= data['Revenue (Millions)'].quantile(0.95))]
get_data
```

## Groupby Operations

```python
director = data.groupby("Director")[["Rating"]].mean()
director.head()
```

## Sorting Operations

```python
sorted_values = director.sort_values(by=['Rating'], ascending=False)
sorted_values
```

## Missing Values

```python
data.isnull().sum()
```

```python
revenue_mean = data['Revenue (Millions)'].mean()
data_indexed.fillna({'Revenue (Millions)': revenue_mean}, inplace=True)
```

```python
def rating_group(rating):
    if rating >= 7.5:
        return "Good"
    elif rating >= 6.0:
        return "Average"
    else:
        return "Bad"

data['Rating_Category'] = data['Rating'].apply(rating_group)

data[["Title", "Director", "Rating", "Rating_Category"]].head(5)
```
